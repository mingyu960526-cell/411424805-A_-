```python
def price(a,b,dis=0.05):
  total=a*b
  tax=total*dis
  total_all=total+tax
  return total,tax,total_all
p=int(input("請輸入單價:"))
q=int(input("請輸入數量:"))
print(f"小記為{price(p,q)[0]}")
print(f"稅金為{price(p,q)[1]}")
print(f"總金額為{price(p,q)[2]}")
```
