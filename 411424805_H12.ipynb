```python
p = int(input("請輸入單價: "))
q = int(input("請輸入數量: "))

def cost():
  return p * q

def cost_9():
  return cost() * 0.9

# 呼叫函數取得計算結果
a = cost()
b = cost_9()

print(f"原價金額:{a}\n折價金額:{b}")
```
