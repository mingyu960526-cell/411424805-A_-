```python
s=7
guess=int(input("請猜一個字"))
while guess !=s:
  if guess>s:
    print("你猜太大了")
  else:
    print("你猜太小了")
  guess=int(input("請猜一個字"))
print("你猜對了")
```

```python
import random
guess = int(input("請猜一個字"))
s=random.randint(1,15)
while guess !=s:
  if guess>s:
    print("你猜太大了")
  else:
    print("你猜太小了")
  guess=int(input("再猜"))
print("猜對了")
```

```python
import random
day =1
max_day=3
win=False
while day <=max_day and not win:
  lucky=random.randint(1,50)
  guess = int(input(f"今天是第{day}天，請輸入1-50數字"))
  if guess ==lucky:
    print("有中")
    win=True
  else:
    if day ==max_day:
      print("很遺憾三天都沒有中獎,下次努力")
      print(f"今天的幸運數字是{lucky}")
      break
    else:
      print("你沒中，明天再來")
      print(f"今天的幸運數字是{lucky}")
      day+=1
```
