```python
order=[]
print('請輸入你要點的餐點，請輸入"/"結束:')
while True:
  item=input()
  if item=="/":
    break
  order.append(item)
  if item=="沙拉":
    order.remove("沙拉")
    print("今日沒有沙拉了，把你的餐點項目踢掉了")
order.append("冰淇凌")
order=",".join(order)
print(f"完整訂單清單:{order}")
```
