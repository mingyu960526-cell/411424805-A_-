```python
a=float(input("請輸入身高(公分)"))
b=float(input("請輸入體重(公斤)"))
BMI=b/((a/100)*(a/100))
print(f"BMI={BMI:.2f}")
if BMI <18.5:
  print("判定為:過輕")
else:
  if BMI<24:
    print("正常")
  else:
    print("判定為:過重")
```
