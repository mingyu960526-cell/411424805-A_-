```python
a = ["木棒", "手槍", "霰彈槍", "狙擊槍", "雷射劍"]
b = [1, 2, 3, 4, 5]

# 修正：將迴圈變數改用 x, y，避免覆寫原本的清單 a 與 b
for x, y in zip(b, a):
  print(x, y)

while True :
  c = int(input("請輸入武器編號(0結束):"))
  if c == 0 :
    break
  if 1 <= c <= len(a):
    # 此時 a 與 b 仍為清單，可以正常用 c-1 索引取值
    print(a[c-1], b[c-1])
  else:
    print("找不到武器")

print("平均攻擊力:", sum(b)/len(b))
```
